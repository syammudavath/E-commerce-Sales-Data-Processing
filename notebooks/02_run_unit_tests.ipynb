# Run unit tests
Runs the pytest suite in `tests/` against `scripts/transformations.py` on the attached compute.
Use it on its own, or via `%run` from `01_ecommerce_sales_pipeline`; the pipeline stops if any test fails.

```python
# Install pytest only if missing. (Not %pip: it restarts Python, which would wipe the
# state of a notebook that calls this one via %run.)
import importlib.util
import subprocess
import sys

if importlib.util.find_spec("pytest") is None:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "pytest>=7.4"])
```

```python
import os
import sys

import pytest

sys.dont_write_bytecode = True  # keep __pycache__ out of the repo
REPO_ROOT = os.path.abspath(os.path.join(os.getcwd(), ".."))
if REPO_ROOT not in sys.path:
    sys.path.insert(0, REPO_ROOT)

# --no-header keeps absolute workspace paths (which contain the user name) out of the output
exit_code = pytest.main([os.path.join(REPO_ROOT, "tests"), "-v", "--no-header",
                         "-p", "no:cacheprovider", "--rootdir", REPO_ROOT])
assert exit_code == 0, f"Unit tests failed (pytest exit code {exit_code})"
print("All unit tests passed")
```
